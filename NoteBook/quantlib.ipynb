# Valuation Date

```python
# ============================================================
# 1. Calibration Date
# ============================================================

calibration_date = "20250102"


# 문자열 YYYYMMDD → QuantLib Date
valuation_date = ql.Date(
    int(calibration_date[6:8]),
    int(calibration_date[4:6]),
    int(calibration_date[0:4])
)


print()
print("==========================================")
print("Calibration Date")
print("==========================================")

print(calibration_date)
print(valuation_date)
```

# BoothStrapping

```python
# ============================================================
# 2. Yield Curve Bootstrapping
# ============================================================

yield_file = (
    r"C:\Python\1.Quantlib\Input\Yield_depositSwap.xlsx"
)


curve, curve_result, curve_info = bootstrap_curve_from_excel(
    yield_file,
    calibration_date
)


print()
print("========================================")
print("Bootstrapped Yield Curve")
print("========================================")

print(
    curve_result.to_string(
        index=False
    )
)

print()
print("Curve Info")
print(curve_info)
```

# Calibration 

```python
# ============================================================
# 3. Swaption Volatility Excel
# ============================================================

swaption_file = (
    r"C:\Python\1.Quantlib\Input\swaption_vol_long_format.xlsx"
)


swaption_data = pd.read_excel(
    swaption_file
)


# ============================================================
# 4. Select Swaption Data by Date
# ============================================================

swaption_data["Data"] = (

    swaption_data["Data"]
    .astype(str)
    .str.replace(".0", "", regex=False)
    .str.strip()

)


swaption_data = swaption_data[
    swaption_data["Data"] == calibration_date
].copy()


if swaption_data.empty:

    raise ValueError(
        f"Excel에 {calibration_date} 날짜의 "
        f"Swaption Vol 데이터가 없습니다."
    )


print()
print("==========================================")
print("Swaption Excel Data")
print("==========================================")

print(
    swaption_data.to_string(
        index=False
    )
)


# ============================================================
# 5. Long Format → Matrix
# ============================================================

swaption_vols = swaption_data.pivot(

    index="옵션만기",

    columns="기초자산만기",

    values="변동성"

)


print()
print("==========================================")
print("Swaption Volatility Matrix")
print("==========================================")

print(
    swaption_vols
)


# ============================================================
# 6. G2++ Calibration
# ============================================================

model, calibration_result, calibration_info = calibrate_g2(
    curve,
    swaption_vols
)
```

# G2++ Monte Carlo Simulation

```python
# ============================================================
# G2++ Monte Carlo Simulation
# ============================================================

simulation = simulate_g2_short_rate(
    curve=curve,
    model=model,
    simulation_years=30,
    n_paths=100000,
    time_step_months=1
)


# ============================================================
# 1. Short Rate Paths
# ============================================================

plot_short_rate_paths(
    simulation,
    n_plot_paths=100000
)


# ============================================================
# 2. Market Zero Curve vs MC Zero Curve
# ============================================================

plot_zero_rate_comparison(
    simulation,
    show_market=True,
    show_mc=True,
    show_band=True
)


# ============================================================
# 3. Zero Rate Error
#    MC Zero Rate - Market Zero Rate
# ============================================================

plot_zero_rate_error(
    simulation
)

# ============================================================
# 4. Zero Rate Comparison Table
# ============================================================

zero_rate_result = print_zero_rate_comparison(
    simulation
)
```

# Bermudan Callable Swap

```python
# ============================================================
# Bermudan Callable Swap
# ============================================================

# ============================================================
# 1. Valuation Date
# ============================================================

valuation_date = valuation_date


# ============================================================
# 2. Trade Date
# ============================================================

trade_date = ql.Date(
    3,
    ql.January,
    2024
)


# ============================================================
# 3. Maturity
# ============================================================

maturity = ql.Date(
    3,
    ql.January,
    2034
)


# ============================================================
# 4. Call Dates
#
# Trade Date ~ Maturity
# 1Y 간격
#
# 단,
# valuation_date 이후의 Call Date만 생성
#
# Trade Date / Maturity 자체는 Call Date에서 제외
# ============================================================

call_dates = generate_call_dates(

    trade_date=trade_date,

    maturity=maturity,

    valuation_date=valuation_date,

    frequency="1Y"

)


# ============================================================
# 5. Call Date 확인
# ============================================================

print_call_dates(
    call_dates
)

# ============================================================
# 6. Bermudan Callable Swap Valuation
# ============================================================

result = price_callable_swap(

    simulation=simulation,

    valuation_date=valuation_date,

    trade_date=trade_date,

    maturity=maturity,

    call_dates=call_dates,

    # --------------------------------------------------------
    # Swap Position
    # --------------------------------------------------------
    # Receiver :
    #   Fixed Receive
    #   Floating Pay
    #
    # Payer :
    #   Fixed Pay
    #   Floating Receive
    # --------------------------------------------------------

    receiver_payer="Receiver",


    # --------------------------------------------------------
    # Call Right
    # --------------------------------------------------------
    # Receiver :
    #   Receiver가 Call 권리 보유
    #
    # Payer :
    #   Payer가 Call 권리 보유
    # --------------------------------------------------------

    call_right="Payer",


    # --------------------------------------------------------
    # Swap Terms
    # --------------------------------------------------------

    fixed_rate=0.0300,

    floating_spread=0.0010,

    notional=10_000,

    fixed_leg_tenor="1Y",

    floating_leg_tenor="6M"

)


# ============================================================
# 7. Result
# ============================================================

print_callable_swap_result(
    result
)
```
